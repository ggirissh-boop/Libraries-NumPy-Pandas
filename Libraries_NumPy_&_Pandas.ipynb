Q1. Load the given retail sales dataset into a Pandas DataFrame and display the first 5 rows along with the dataset information to understand its structure.

```python
import pandas as pd
df = pd.read_csv('/content/drive/MyDrive/Libraries: NumPy & Pandas/retail_dataset .csv')
print(df.head())
print(df.info())
```

Q2. Inspect the dataset and identify missing values in each column. Display the total number of missing values present in the dataset.

```python
print("Missing values per column:")
print(df.isnull().sum())

print("\nTotal missing values in the dataset:")
print(df.isnull().sum().sum())
```

Q3. Replace missing values in the Quantity and Price columns with their respective mean values using Pandas.

```python
# Calculate the mean for 'Quantity' and 'Price' columns
mean_quantity = df['Quantity'].mean()
mean_price = df['Price'].mean()

# Replace missing values with the calculated means
df['Quantity'] = df['Quantity'].fillna(mean_quantity)
df['Price'] = df['Price'].fillna(mean_price)

# Display the updated missing values to confirm imputation
print("Missing values after imputation:")
print(df.isnull().sum())
```

Q4. Remove all rows where the Product Category or Region column contains missing values.

```python
# Remove rows where 'Product Category' or 'Region' columns have missing values
df.dropna(subset=['Product Category', 'Region'], inplace=True)

# Display the updated missing values to confirm removal
print("Missing values after removing rows with missing 'Product Category' or 'Region':")
print(df.isnull().sum())

# Display the new shape of the DataFrame
print(f"\nNew DataFrame shape: {df.shape}")
```

Q5. Create a new column called Revenue using NumPy, where:
Revenue = Quantity × Price
Display the updated dataset with the new column.

```python
import numpy as np

# Create the 'Revenue' column using NumPy's multiply function
df['Revenue'] = np.multiply(df['Quantity'], df['Price'])

# Display the updated dataset with the new 'Revenue' column
print("Dataset with new 'Revenue' column:")
display(df.head())
```

Q6. Using NumPy, calculate the total revenue generated from all transactions in the dataset.

```python
total_revenue = np.sum(df['Revenue'])
print(f"Total Revenue: ${total_revenue:,.2f}")
```

Q7. Group the dataset by Product Category and calculate the total revenue generated by each category.

```python
category_revenue = df.groupby('Product Category')['Revenue'].sum().reset_index()
print("Total Revenue by Product Category:")
display(category_revenue.round(2))
```

Q8. Based on the revenue calculated for each product category, identify the top 3 categories with the
highest revenue and the bottom 3 categories with the lowest revenue.

```python
# Sort by Revenue to find top and bottom categories
sorted_category_revenue = category_revenue.sort_values(by='Revenue', ascending=False).round(2)

# Top 3 categories
top_3_categories = sorted_category_revenue.head(3)
print("Top 3 Product Categories by Revenue:")
display(top_3_categories)

print("\n")

# Bottom 3 categories
bottom_3_categories = sorted_category_revenue.tail(3)
print("Bottom 3 Product Categories by Revenue:")
display(bottom_3_categories)
```

Q9. Group the dataset by Region and calculate the total revenue generated in each region. Identify the
region with the highest revenue and the region with the lowest revenue.

```python
region_revenue = df.groupby('Region')['Revenue'].sum().reset_index()
region_revenue = region_revenue.round(2)

print("Total Revenue by Region:")
display(region_revenue)

# Identify region with highest revenue
highest_revenue_region = region_revenue.loc[region_revenue['Revenue'].idxmax()]
print(f"\nRegion with Highest Revenue:\n{highest_revenue_region['Region']} with ${highest_revenue_region['Revenue']:,.2f}")

# Identify region with lowest revenue
lowest_revenue_region = region_revenue.loc[region_revenue['Revenue'].idxmin()]
print(f"\nRegion with Lowest Revenue:\n{lowest_revenue_region['Region']} with ${lowest_revenue_region['Revenue']:,.2f}")
```

Q10. Extract the month from the Date column using Pandas and group the dataset by month to calculate the
total monthly revenue. Also, calculate the mean, median, and standard deviation of the Revenue column
using NumPy.

```python
import numpy as np

# Ensure 'Date' column is in datetime format
df['Date'] = pd.to_datetime(df['Date'], format='%d-%m-%Y', errors='coerce')

# Extract month from the 'Date' column
df['Month'] = df['Date'].dt.month

# Group by month and calculate total monthly revenue
monthly_revenue = df.groupby('Month')['Revenue'].sum().reset_index()
print("Total Monthly Revenue:")
display(monthly_revenue.round(2))

# Calculate mean, median, and standard deviation of the Revenue column using NumPy
mean_revenue = np.mean(df['Revenue'])
median_revenue = np.median(df['Revenue'])
std_dev_revenue = np.std(df['Revenue'])

print(f"\nMean Revenue: ${mean_revenue:,.2f}")
print(f"Median Revenue: ${median_revenue:,.2f}")
print(f"Standard Deviation of Revenue: ${std_dev_revenue:,.2f}")
```
